<h1><font color="#004D7F" size=5>&#127760;Qué és la Bioinformàtica?</font></h1>

*   Es podria definir com l'us de la informàtica per a recollir,
manipular, analitzar, investigar i distribuir dades biològiques
de qualsevol tipus.

*   Hauríem d'entendre la bioinformàtica no tant com un nou
camp de la biologia, sinó més aviat com una nova manera de
fer biologia.



![Bioiformatics](https://www.researchgate.net/publication/282040422/figure/fig12/AS:295139462402058@1447378139528/Components-of-bioinformatics-Genomics-is-a-branch-of-science-studying-structural-and.png)

![Bioiformatics](https://t4.ftcdn.net/jpg/17/28/34/85/240_F_1728348507_09dF8fT0ebXqeXm0aTDeAidVB4x7hq9a.jpg)


Bioinformatics:
* És un camp de la ciència interdisciplinari que desenvolupa mètodes i eines de programari per entendre les dades biològiques, especialment quan els conjunts de dades són grans i complexos.
* La bioinformàtica utilitza la biologia, la química, la física, la informàtica, la programació informàtica, l'enginyeria de la informació, les matemàtiques i l'estadística per analitzar i interpretar dades biològiques.
* El procés posterior d'anàlisi i interpretació de dades s'anomena biologia computacional.



**UNA DISCIPLINA AMB MOLTA PROJECCIÓ**

* La Bioinformàtica és una de les disciplines científiques que més protagonisme i projecció estan tenint en els darrers anys, cosa que encara és més visible, a partir de l' any 2020 amb la seva tasca fonamental en el maneig i interpretació de dades sobre el SARS-CoV-2.

* La seva tasca consisteix a investigar, desenvolupar i aplicar eines informàtiques i computacionals per permetre i millorar el maneig de dades biològiques, gràcies a l'ús de eines que reuneixen, emmagatzemar, organitzar, analitzen i permeten interpretar aquestes dades.

* El seu creixement va ser lligat al desenvolupament de la Biologia Molecular, el descobriment de l'ADN, i dels avenços en computació.

<h1><font color="#004D7F" size=5>&#127760;Objectius de la Bioinformàtica</font></h1>

* Esperem que els avenços tecnològics i la gran quantitat d'informació disponible
ens permetin aproximacions globals als problemes de la biologia actual (com ara
simulacions cel·lulars completes, comprensió global de vies gèniques, xarxes
d'interaccions proteiques, ...).
* Aquests avenços han de portar-nos, un altre cop, de la simple recollida de dades al
postulat d'hipòtesis. Per tant, cal un canvi del paradigma actual, passant de l'anàlisi
(aproximacions reduccionistes) a la síntesi (aproximacions integratives).


<h1><font color="#004D7F" size=5>&#127760;Alguns camps de la bioinformàtica</font></h1>

Històricament, la bioinformàtica s'ha basat en els apartats principals: 1)Anàlisi de seqüències, 2)Biologia estructural, 3)Estudi de gens i enginyeria genètica i 4)Biologia computacional. Durant els últims 20 anys, la Bioinformàtica s'ha transformat completament, i avui és impossible parlar de 2 a 5 àrees principals de la bioinformàtica, sino de moltes mes. Algunes son:

* Anotació de seqüències (reconstrucció de gens/genomes, predicció de regions codificants i/o
reguladores, anàlisi de motius, ...)
* Anotació de genomes (acoblat de genomes, creació de bases de dades, eines d'accés a les dades
i manteniment, ...)
* Biologia evolutiva
* Anàlisi de l'expressió gènica
* Anàlisi de la variabilitat
* Anàlisi de biodiversitat
* Predicció de l'estructura de proteïnes
* Modelat de sistemes biològics
* Anàlisi d'imatges

<h2><font color="#004D7F" size=5>&#127760;Origen de la Bioinformàtica</font></h2>

![Bioiformatics](https://stpperspectives.files.wordpress.com/2018/08/bioinformatics-timeline-2.png)

* És fàcil per als estudiants i investigadors d'avui creure que la bioinformàtica moderna va sorgir recentment per ajudar a l'anàlisi de dades de seqüenciació de nova generació.
* Tanmateix, els inicis de la bioinformàtica es van produir fa més de 50 anys, quan els ordinadors de sobretaula encara eren una hipòtesi i l'ADN encara no es podia seqüenciar.
* Les bases de la bioinformàtica es van establir a principis dels anys 60 amb l'aplicació de mètodes computacionals a l'anàlisi de seqüències de proteïnes (en particular, l'assemblatge de seqüències de novo, bases de dades de seqüències biològiques i models de substitució).
* Més tard, l'anàlisi d'ADN també va sorgir a causa dels avenços paral·lels en (i) mètodes de biologia molecular, que van permetre una manipulació més fàcil de l'ADN, així com la seva seqüenciació, i (ii) la informàtica, que va veure l'aparició d'ordinadors cada cop més miniaturitzats i més potents, així com un nou programari més adequat per gestionar tasques de bioinformàtica.
* Entre els anys 90 i els 2000, les grans millores en la tecnologia de seqüenciació, juntament amb la reducció de costos, van donar lloc a un augment exponencial de dades.
* L'arribada del 'Big Data' ha plantejat nous reptes en termes de mineria i gestió de dades, que demana més coneixements de la informàtica al camp.
* Juntament amb una quantitat cada cop més gran d'eines bioinformàtiques, el Big Data biològic va tenir (i continua tenint) implicacions profundes en el poder predictiu i la reproductibilitat dels resultats de la bioinformàtica.
* Per superar aquest problema, ara les universitats estan integrant plenament aquesta disciplina al pla d'estudis dels estudiants de biologia.
* Subdisciplines recents com la biologia sintètica, la biologia de sistemes i el modelatge de cèl·lules senceres han sorgit de la complementarietat cada cop més creixent entre la informàtica i la biologia.
